## 实验2：函数、条件、循环与递归

```python
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("分数必须在0到100之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

print(score_level(95), score_level(85), score_level(70), score_level(50))

sum_for = 0
for i in range(1, 101):
    sum_for += i
print("for循环求和:", sum_for)

sum_while = 0
j = 1
while j <= 100:
    sum_while += j
    j += 1
print("while循环求和:", sum_while)
```

```python
def fact_recursive(n):
    if n == 0:
        return 1
    return n * fact_recursive(n - 1)

def fact_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print("递归 n=5:", fact_recursive(5), "迭代 n=5:", fact_iterative(5))

def summary(values):
    return (min(values), max(values), sum(values) / len(values))

min_val, max_val, avg_val = summary([10, 20, 30])
print("summary解包:", min_val, max_val, avg_val)

def sum_all(*args):
    return sum(args)

print("无参数:", sum_all())
print("单个参数:", sum_all(5))
print("多个参数:", sum_all(1, 2, 3, 4))
```

## 实验3：列表、字典、元组与基础统计

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

score_dict = dict(zip(names, scores))
for name, score in score_dict.items():
    print(f"姓名: {name}, 成绩: {score}")

for index, name in enumerate(names, start=1):
    print(f"序号 {index}: {name} - {scores[index-1]}")

passed_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [s * s for s in scores]
print("及格学生:", passed_names)
print("成绩平方:", squared_scores)

count = len(scores)
avg = sum(scores) / count
max_score = max(scores)
min_score = min(scores)
pass_rate = len(passed_names) / count

print(f"总人数: {count}")
print(f"平均分: {avg:.2f}")
print(f"最高分: {max_score}, 最低分: {min_score}")
print(f"及格率: {pass_rate:.1%}")
```

*(任务4要求演示深浅拷贝，每次比较都从新建的原始对象开始)*

```python
import copy

print("直接赋值")
obj1 = {"name": "Alice", "scores": [85, 90]}
obj2 = obj1
obj2["scores"].append(100)
print("原对象:", obj1)

print("浅拷贝")
obj3 = {"name": "Bob", "scores": [75, 80]}
obj4 = copy.copy(obj3)
obj4["scores"].append(100)
print("原对象:", obj3)

print("深拷贝")
obj5 = {"name": "Carol", "scores": [95, 88]}
obj6 = copy.deepcopy(obj5)
obj6["scores"].append(100)
print("原对象:", obj5)
```
